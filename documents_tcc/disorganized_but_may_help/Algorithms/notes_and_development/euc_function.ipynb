```python
def calculate_distance_matrix(points_2d):
    n = len(points_2d)
    D = [[0 for _ in range(n)] for _ in range(n)]

    for i in range(n):
        for j in range(n):
            if i != j:
                # Extract coordinates (ignoring the node identifier)
                x1, y1 = points_2d[i][1], points_2d[i][2]
                x2, y2 = points_2d[j][1], points_2d[j][2]
                distance = math.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)
                D[i][j] = distance
            else:
                D[i][j] = 0  # Distance to itself is 0
    return D


# Example usage
distance_matrix = calculate_distance_matrix(points_2d)
```

```python
def calculate_distance_matrix(points):
    # Create a dictionary to store distances with node indices as keys
    distance_matrix = {node[0]: {} for node in points}  # Initialize with node indices

    for i in range(len(points)):
        node1, x1, y1 = points[i]
        for j in range(len(points)):
            node2, x2, y2 = points[j]
            if node1 != node2:
                distance = math.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)
                distance_matrix[node1][node2] = distance
            else:
                distance_matrix[node1][node2] = 0  # distance to itself is 0
    return distance_matrix


# Example usage
distance_matrix_dict = calculate_distance_matrix(points_2d)
```

```python
df = pd.DataFrame(distance_matrix_dict)
df
```

```python
def euc_3d(data: list[tuple[int]], problem_type: str = "TSP") -> dict:
    """
    Calculate the Euclidean distance between two points in 3D space.
    """
    distance_matrix = {node[0]: {} for node in data}

    for i, (node_i, x_i, y_i, z_i) in enumerate(data):
        # If the problem is symmetric
        if problem_type == "TSP":
            distance_matrix[node_i][node_i] = 0
            for j in range(i + 1, len(data)):
                node_j, x_j, y_j, z_j = data[j]
                distance = math.sqrt(
                    (x_i - x_j) ** 2 + (y_i - y_j) ** 2 + (z_i - z_j) ** 2
                )
                distance_matrix[node_i][node_j] = distance
                distance_matrix[node_j][node_i] = distance

        else:
            for j, (node_j, x_j, y_j, z_j) in enumerate(data):
                if node_i != node_j:
                    distance = math.sqrt(
                        (x_i - x_j) ** 2 + (y_i - y_j) ** 2 + (z_i - z_j) ** 2
                    )
                    distance_matrix[node_i][node_j] = distance
                else:
                    distance_matrix[node_i][node_j] = 0

    return distance_matrix


distance_matrix_3d_dict = euc_3d(points_3d, problem_type="TSP")
```
